# NSL-KDD Preprocessing

Produces `data/processed/{train,test}.pt` with shape `(N, 120)` per `plan.md`.

- 41 raw columns, drop `difficulty`
- Drop near-constant `num_outbound_cmds`, `is_host_login`
- One-hot encode `protocol_type`, `service`, `flag` (union of train+test categories)
- Min-max scale
- Two label views: binary (normal vs attack), 5-class (Normal/DoS/Probe/R2L/U2R)

```python
import sys
sys.path.insert(0, 'src')
import numpy as np
import pandas as pd
from nids_dl.data import load_raw, DROP_CONSTANT, ATTACK_FAMILY
```

```python
train_df = load_raw('NSL-KDD/KDDTrain+.txt')
test_df = load_raw('NSL-KDD/KDDTest+.txt')
print('train', train_df.shape, 'test', test_df.shape)
train_df.head()
```

## Label distributions

```python
print('binary (train):')
print((train_df['label'] != 'normal').value_counts())
print('\nbinary (test):')
print((test_df['label'] != 'normal').value_counts())
```

```python
def fam(lbl):
    return ATTACK_FAMILY.get(lbl, 'R2L')
print('5-class (train):')
print(train_df['label'].map(fam).value_counts())
print('\n5-class (test):')
print(test_df['label'].map(fam).value_counts())
```

## Verify constant columns to drop

```python
for col in DROP_CONSTANT:
    print(col, 'train nunique=', train_df[col].nunique(), 'test nunique=', test_df[col].nunique())
```

## Run preprocessor and save tensors

```python
from nids_dl.data import build_and_save
info = build_and_save('NSL-KDD/KDDTrain+.txt', 'NSL-KDD/KDDTest+.txt', 'data/processed')
print('train', info['train_shape'])
print('test', info['test_shape'])
print('n_features', info['n_features'])
assert info['n_features'] == 120, f"expected 120 features, got {info['n_features']}"
```

```python
import torch
d = torch.load('data/processed/train.pt', weights_only=False)
print('X', d['X'].shape, d['X'].dtype)
print('y_bin', d['y_bin'].shape, 'classes', torch.unique(d['y_bin']).tolist())
print('y_mul', d['y_mul'].shape, 'classes', torch.unique(d['y_mul']).tolist())
print('min/max', d['X'].min().item(), d['X'].max().item())
```
